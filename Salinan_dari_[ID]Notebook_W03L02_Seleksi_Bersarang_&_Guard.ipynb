## 3. Explain: Pola Guard & Validasi Masukan (.isdigit())

### Mengapa Perlu Guard?
Ketika kita memanggil fungsi `int(input())` secara langsung, program akan *crash* (`ValueError`) jika pengguna memasukkan karakter non-angka seperti `"dua"` atau `"-3"` (karena tanda minus tidak dikenali oleh `.isdigit()`).

`text.isdigit()` mengembalikan `True` jika dan hanya jika semua karakter di dalam teks adalah angka (0-9).

### Pola Guard dengan .isdigit():

```python
umur_teks = input("Umur: ")

# 1. Guard Clause (Memeriksa validitas masukan)
if not umur_teks.isdigit():
    print("Masukan harus angka bulat")
else:
    # 2. Proses data jika sudah aman
    umur = int(umur_teks)
    if umur > 120:
        print("Umur tidak masuk akal")
    elif umur >= 17:
        print("Boleh membuat KTP")
    else:
        print("Belum cukup umur")
```

## 4. Elif Berantai vs If Bersarang (U3.5)

* **`elif` Berantai**:
  Digunakan untuk membandingkan **satu variabel** dengan beberapa rentang kondisi yang tidak saling tumpang tindih.
* **`if` Bersarang**:
  Digunakan jika suatu kondisi baru bernilai relevan dan ingin diperiksa **setelah kondisi luar dipastikan terpenuhi** (kondisi saling bergantung).
