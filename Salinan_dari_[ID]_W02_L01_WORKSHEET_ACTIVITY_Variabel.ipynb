# 3. Explain: ekspresi dan statement

- **Ekspresi** adalah gabungan nilai, variabel, dan/atau operator yang menghasilkan sebuah nilai. Contoh: `harga * jumlah`.
- **Statement** adalah instruksi yang dijalankan program. Contoh: `total = harga * jumlah` dan `print(total)`.

Pada statement penugasan, bagian di sebelah kanan tanda `=` adalah ekspresi. Seluruh baris penugasan adalah statement.

> **Catatan Jupyter:** sel terakhir yang hanya berisi ekspresi dapat menampilkan nilainya otomatis. Dalam program Python biasa, gunakan `print()` bila Anda ingin menampilkan nilai ke layar.


````python
harga = 12000                 # statement penugasan
jumlah = 3                    # statement penugasan
harga * jumlah                # ekspresi; Jupyter menampilkan nilainya karena berada di baris terakhir
````

````python
total = harga * jumlah        # bagian kanan adalah ekspresi; seluruh baris adalah statement
print("Total:", total)      # statement untuk menampilkan keluaran
````
